### Bước 8. Đếm giá trị `unknown` ở các thuộc tính định tính
Bộ dữ liệu này không để trống ô mà ghi giá trị thiếu là `unknown`, nên `info()` báo không thiếu dữ liệu. Cần đếm riêng để biết thực tế thuộc tính nào bị thiếu.

```python
unknown = (df.select_dtypes(exclude='number') == 'unknown').sum()
unknown_pct = (unknown / len(df) * 100).round(2)
pd.concat([unknown, unknown_pct], axis=1, keys=['Số unknown', '%']).sort_values('Số unknown', ascending=False)
```
