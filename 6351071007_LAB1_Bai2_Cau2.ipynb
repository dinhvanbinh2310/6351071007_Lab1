### Bước 1. Import thư viện và đọc dữ liệu

```python
import numpy as np
import pandas as pd

with open('titanic.csv') as f:
    df = pd.read_csv(f, dtype={'age': np.float64, 'sibsp': np.int64, 'parch': np.int64})

print('Kích thước dữ liệu (dòng, cột):', df.shape)
```

### Bước 2. Đếm số giá trị rỗng và tỷ lệ % của từng thuộc tính
- `df.isnull().sum()`: số giá trị rỗng của mỗi cột.
- `df.isnull().count()`: tổng số dòng của mỗi cột (1309) → chia ra tỷ lệ %.
- Sắp xếp giảm dần rồi ghép 2 cột `Total` và `%` lại.

```python
total = df.isnull().sum().sort_values(ascending=False)
percent_1 = df.isnull().sum() / df.isnull().count() * 100
percent_2 = (round(percent_1, 1)).sort_values(ascending=False)
missing_data = pd.concat([total, percent_2], axis=1, keys=['Total', '%'])
missing_data.head(5)
```

### Bước 3. Danh sách 5 thuộc tính thiếu dữ liệu nhiều nhất

```python
top5 = missing_data.head(5).index.tolist()
print('5 thuộc tính có nhiều giá trị rỗng nhất:', top5)
```

### Nhận xét
- 5 thuộc tính thiếu dữ liệu nhiều nhất: **body** (1188 – 90.8%), **cabin** (1014 – 77.5%), **boat** (823 – 62.9%), **home.dest** (564 – 43.1%), **age** (263 – 20.1%).
- Thuộc tính `body` thiếu dữ liệu rất nghiêm trọng (90.8%), điều này sẽ gây ảnh hưởng lớn nếu sử dụng thuộc tính này trong quá trình khai thác → nên loại bỏ.
- `cabin` và `boat` cũng thiếu hơn một nửa nên khó điền giá trị thay thế chính xác.
- `age` chỉ thiếu khoảng 20% và là thuộc tính quan trọng → nên điền giá trị thiếu (trung bình/trung vị) thay vì xóa.
