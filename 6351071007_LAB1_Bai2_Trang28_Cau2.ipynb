### Bước 1. Import thư viện và đọc dữ liệu

```python
import numpy as np
import pandas as pd

with open('bank-additional-full.csv') as f:
    df = pd.read_csv(f, sep=';')

print('Kích thước dữ liệu (dòng, cột):', df.shape)
```

### Bước 2. Kiểm tra giá trị rỗng theo cách thông thường
Nếu chỉ dùng `isnull()` thì bộ dữ liệu này **không có ô nào rỗng**:

```python
print('Tổng số ô rỗng (NaN):', df.isnull().sum().sum())
```

### Bước 3. Chuyển giá trị `unknown` thành giá trị rỗng
Theo mô tả của bộ dữ liệu (UCI), các thuộc tính định tính bị thiếu được ghi là `unknown`. Vì vậy cần đổi `unknown` thành `NaN` thì mới đếm đúng số giá trị rỗng.

```python
df = df.replace('unknown', np.nan)
print('Tổng số ô rỗng sau khi đổi unknown → NaN:', df.isnull().sum().sum())
```

### Bước 4. Đếm số giá trị rỗng và tỷ lệ % của từng thuộc tính
Làm giống hướng dẫn Câu 2 của Titanic: đếm, tính %, sắp xếp giảm dần rồi lấy 5 thuộc tính đầu.

```python
total = df.isnull().sum().sort_values(ascending=False)
percent_1 = df.isnull().sum() / df.isnull().count() * 100
percent_2 = (round(percent_1, 2)).sort_values(ascending=False)
missing_data = pd.concat([total, percent_2], axis=1, keys=['Total', '%'])
missing_data.head(5)
```

### Bước 5. Danh sách 5 thuộc tính thiếu dữ liệu nhiều nhất

```python
top5 = missing_data.head(5).index.tolist()
print('5 thuộc tính có nhiều giá trị rỗng nhất:', top5)
```

### Bước 6. Các thuộc tính còn lại có bị thiếu không?

```python
missing_data[missing_data['Total'] > 0]
```
